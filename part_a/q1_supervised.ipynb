```python
# Step 1 - Data Loading and Inspection
import pandas as pd

# Load dataset
df = pd.read_csv(r"C:\Users\yasmi\OneDrive\Documents\ml-assessment-yasminbano-khan\data\q1_heart_disease.csv")

# Shape (rows, columns)
print("Shape:", df.shape)

# Data types
print("\nData Types:\n", df.dtypes)

# Missing values
print("\nMissing Values:\n", df.isnull().sum())

# First 5 rows
df.head()

# Step 2 —  Exploratory Data Analysis

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='heart_disease', data=df)
plt.title("Heart Disease Count")
plt.show()

# Interpretation - From the graph we can compare patient with and without heart disease.

plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# Interpretation - The heatmap shows different features are co related with each other.

sns.histplot(df['age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()

# Interpretation - This plot shows the distribution of patient ages which help us indicate which age group is most or less affected.

# Step 3 - Data Preprocessing

# Handle missing value
df.fillna(df.median(numeric_only=True), inplace=True)

# Convert categorical to numbers
df = pd.get_dummies(df, drop_first=True)

# Split X and y
X = df.drop('heart_disease', axis=1)
y = df['heart_disease']

# Scale data
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, stratify=y, random_state=42
)
print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)

# Step 4 — Model Training

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# Create models
dt = DecisionTreeClassifier(random_state=42)
rf = RandomForestClassifier(random_state=42)
gb = GradientBoostingClassifier(random_state=42)

# Train models
dt.fit(X_train, y_train)
rf.fit(X_train, y_train)
gb.fit(X_train, y_train)

print("All models trained successfully!")

# Step 5 — Model Evaluation

from sklearn.metrics import confusion_matrix, classification_report

models = {
    "Decision Tree": dt,
    "Random Forest": rf,
    "Gradient Boosting": gb
}

for name, model in models.items():
    print(f"\n{name}")
    
    y_pred = model.predict(X_test)
    
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
    print("Classification Report:\n", classification_report(y_test, y_pred))

```

1. More patients are in class 1 (disease present) than 0 (no disease). This shows the dataset is slightly imbalanced but fairly balanced overall.
2. Some features like max_hr and chest_pain_type show a strong relationship with heart_disease, while features like oldpeak and st_slope show noticeable positive or negative correlations.
3. Most patients fall between the age of 40 and 60 years, indicating that middle-aged individuals form the majority of the dataset.


1. Handle Missing Values - Used median imputation to handle missing values because it is less affected by outliers and keeps the data stable.
2. Convert text to numbers - Used one-hot encoding to convert categorical variables into numerical format so that the models can process them.
4. Scale data - Scaled the data using StandardScaler so that all features have similar ranges.
5. Train-test split - Used stratified splitting to maintain equal classes in both training and testing datasets.

1. Random Forest model performs best because it has a higher F1-score and a better balance between precision and recall compared to the other models. This indicates more reliable performance across both classes.

1. After tuning, the model improved because the F1-score increased, indicating better balance between precision and recall.
The best parameters were n_estimators = 100 and max_depth = 5.
